```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df_train = pd.read_csv('../data/processed/feature_engineered_train.csv')
df_eval = pd.read_csv('../data/processed/feature_engineered_eval.csv')
```

```python
target = 'price'
X_train = df_train.drop(columns=[target])
y_train = df_train[target]

X_eval = df_eval.drop(columns=[target])
y_eval = df_eval[target]
```

## Scaler

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_eval_scaled = scaler.transform(X_eval)
```

## Linear Regression

```python
lr = LinearRegression()
lr.fit(X_train_scaled, y_train)
y_pred_lr = lr.predict(X_eval_scaled)

mae = mean_absolute_error(y_eval.values, y_pred_lr)
rmse = np.sqrt(mean_squared_error(y_eval.values, y_pred_lr))
r2 = r2_score(y_eval.values, y_pred_lr)

print('Linear Regression\n')
print(f'MAE: {mae:,.2f}')
print(f'RMSE: {rmse:,.2f}')
print(f'R2: {r2:,.4f}')
```
