```python
import boto3
from pathlib import Path


# Configuration
bucket = 'housemarket-ml-end-to-end'
region = 'eu-west-2'

ROOT_PATH = Path.cwd().parents[0]
DATA_PATH = ROOT_PATH / 'data/'
MODELS_PATH = ROOT_PATH / 'models/'

s3 = boto3.client("s3", region_name=region)

def upload_file(local_path: Path, s3_path: str):
    if not local_path.exists():
        print(f'❌ file does not exist : {local_path}')
    else:
        s3.upload_file(local_path, bucket, s3_path)
        print(f'⬆️ {local_path} uploaded to the {bucket} bucket')

    return None

# Datasets

ROOT_PATH = Path('./').resolve().parents[0]
DATA_PATH = ROOT_PATH / 'data'
PREPROCESSED_HOLDOUT_PATH = DATA_PATH / 'processed' / 'preprocessed_holdout.csv'
FE_HOLDOUT_PATH = DATA_PATH / 'feature_engineered' / 'fe_holdout.csv'

upload_file(PREPROCESSED_HOLDOUT_PATH, 'preprocessed_holdout.csv')
upload_file(FE_HOLDOUT_PATH, 'fe_holdout.csv')
# upload_file(DATA_PATH / 'raw/holdout.csv', 'processed/holdout_csv')

# # Model
# upload_file(MODELS_PATH / 'best_xgb_model.pkl', 'models/best_xgb_model.pkl')

# def upload_file(s3):
```
