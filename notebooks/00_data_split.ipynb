```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df = df.sort_values(by=['date'])
```

Separation of the data in three different datasets : 
- 'train.csv' 
- 'eval.csv'
- 'holdout.csv'

```python
df.shape
```

```python
cutoff_eval = '2020-01-01'
cutoff_holdout = '2022-01-01'
```

```python
df_train = df[df['date'] <= cutoff_eval]
df_eval = df[(df['date'] > cutoff_eval) & (df['date'] <= cutoff_holdout)]
df_holdout = df[df['date'] > cutoff_holdout]
```

```python
for _df in [df_train, df_eval, df_holdout]:
    print(_df.shape)
    print(_df.shape[0] / df.shape[0])
    print(' ')
```

```python
df_train.to_csv('../data/raw/train.csv', index=False)
df_eval.to_csv('../data/raw/eval.csv', index=False)
df_holdout.to_csv('../data/raw/holdout.csv', index=False)
```
