```python
import pandas as pd
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
df_train = pd.read_csv('../data/processed/feature_engineered_train.csv')
df_eval = pd.read_csv('../data/processed/feature_engineered_eval.csv')
```

```python
df_train
```

```python
target = 'price'
X_train = df_train.drop(columns=[target])
y_train = df_train[target]

X_eval = df_eval.drop(columns=[target])
y_eval = df_eval[target]
```

```python
y_train.head()
y_eval.head()
```

```python
baseline = DummyRegressor(strategy='median') # always predict the median of the training set
baseline.fit(X_train, y_train)
```

```python
y_prediction = baseline.predict(X_eval)

mae = mean_absolute_error(y_eval.values, y_prediction)
rmse = np.sqrt(mean_squared_error(y_eval.values, y_prediction))
r2 = r2_score(y_eval.values, y_prediction)

print('Baseline Dummy Regressor (Median)\n')
print(f'MAE: {mae:,.2f}')
print(f'RMSE: {rmse:,.2f}')
print(f'R2: {r2:,.4f}')
```
