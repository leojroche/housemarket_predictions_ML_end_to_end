```python
import pandas as pd
from category_encoders import TargetEncoder

df_train = pd.read_csv('../data/processed/cleaned_train.csv')
df_eval = pd.read_csv('../data/processed/cleaned_eval.csv')
# df_holdout = pd.read_csv('../data/raw/holdout')

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
```

```python
df_train.head()
```

## 'date' -> 'year', 'quart', 'month'

```python
df_train['date'] = pd.to_datetime(df_train['date'])
df_train['year'] = df_train['date'].dt.year
df_train['quart'] = df_train['date'].dt.quarter
df_train['month'] = df_train['date'].dt.month

df_eval['date'] = pd.to_datetime(df_eval['date'])
df_eval['year'] = df_eval['date'].dt.year
df_eval['quart'] = df_eval['date'].dt.quarter
df_eval['month'] = df_eval['date'].dt.month
```

```python
df_train.head()
```

## 'zipcode' frequency encodeding

```python
zipcode_mapping = df_train['zipcode'].value_counts()

zipcode_mapping
```

```python
df_train['zipcode_freq'] = df_train['zipcode'].map(zipcode_mapping)

df_eval['zipcode_freq'] = df_eval['zipcode'].map(zipcode_mapping).fillna(0)
# df_eval = df_eval[df_eval['zipcode_freq'].isnull() == False]
df_eval[df_eval['zipcode_freq'].isnull()]
```

## Target encoding ('city_full' and 'price')

```python
te = TargetEncoder(cols=['city_full'])

df_train['city_encoded'] = te.fit_transform(df_train['city_full'], df_train['price'])
df_eval['city_encoded'] = te.transform(df_eval['city_full'])

print(df_train[['city_full', 'city_encoded']].head())
```

## Drop unused columns

```python
col_drops = ['date', 'zipcode', 'city_full', 'city', 'median_sale_price']
df_train.drop( columns = col_drops, inplace=True)
df_eval.drop( columns = col_drops, inplace=True)

df_train.head()
df_eval.head()
```

## Save featured-engineer datasets

```python
df_train.to_csv('../data/processed/feature_engineered_train.csv', index=False)
df_eval.to_csv('../data/processed/feature_engineered_eval.csv', index=False)
```
